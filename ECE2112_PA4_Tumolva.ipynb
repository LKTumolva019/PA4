```python
#Import panda library and load file
import pandas as pd

# Use pandas to read the CSV file into a DataFrame
df = pd.read_csv('board2.csv')
df
```

# A. VISAYAS COMMUNICATION DATAFRAME

```python
#Adding Average Column
# Add 'average' column
df['Average'] = df[['Math', 'Electronics', 'GEAS', 'Communication']].mean(axis=1)
df
```

```python
# Create a DataFrame named VisComm containing students whose Hometown is Visayas and whose Track is Communication.
VisComm = df.loc[(df['Hometown'] == 'Visayas') & (df['Track'] == 'Communication'), ['Name', 'Gender', 'Math', 'Electronics', 'Average']]
```

```python
# Getting the number of rows
rows, columns = VisComm.shape
print("Number of Rows:", rows)

VisComm
```

# C. CATEGORY-AVERAGE VISUALIZATION

### a. For each feature, compute the mean of Average for every category using Pandas.

```python
import matplotlib.pyplot as plt
gender = df.groupby('Gender')
track = df.groupby('Track')
hometown = df.groupby('Hometown')

mean_gender = gender.mean(numeric_only=True)
mean_track = track.mean(numeric_only=True)
mean_hometown = hometown.mean(numeric_only=True)
```

### b. Display the three summary tables.

```python
display(mean_track)
display(mean_gender)
display(mean_hometown)
```

### c. Create one figure containing three bar charts: mean Average by Track, by Gender, and by Hometown.

```python
# TEST MUNA
# Create one single row layout containing three side-by-side plots
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Mean Average by Track
mean_track['Average'].plot(kind='bar', ax=axes[0], color='cornflowerblue')
axes[0].set_title('Mean Average by Track')
axes[0].set_xlabel('Track')
axes[0].set_ylabel('Mean Average')


# Mean Average by Gender
mean_gender['Average'].plot(kind='bar', ax=axes[1], color='lightcoral')
axes[1].set_title('Mean Average by Gender')
axes[1].set_xlabel('Gender')


# Mean Average by Hometown
mean_hometown['Average'].plot(kind='bar', ax=axes[2], color='mediumaquamarine')
axes[2].set_title('Mean Average by Hometown')
axes[2].set_xlabel('Hometown')


# Displaying the charts
plt.tight_layout()
plt.show()
```
